```python
import os
from google import genai
import ipywidgets as widgets
from IPython.display import display, Markdown

# 1. Initialize the client (it automatically picks up the GEMINI_API_KEY env variable)
client = genai.Client()

# 2. Start a chat session using the recommended default model
chat = client.chats.create(model="gemini-3.1-flash-lite")

# 3. Create UI elements for the notebook
output_area = widgets.Output()
text_input = widgets.Text(
    value='',
    placeholder='Type your message here...',
    description='You:',
    disabled=False,
    layout=widgets.Layout(width='70%')
)

def on_submit(sender):
    user_message = text_input.value
    if not user_message.strip():
        return
    
    # Clear the input box immediately
    text_input.value = ''
    
    with output_area:
        # Display user message
        display(Markdown(f"**You:** {user_message}"))
        
        try:
            # Send message to the model (history is managed automatically by the chat object)
            response = chat.send_message(user_message)
            display(Markdown(f"**Gemini:** {response.text}"))
        except Exception as e:
            display(Markdown(f"**Error:** {str(e)}"))
            
        display(Markdown("---"))

# Bind the submit event (pressing Enter) to the function
text_input.on_submit(on_submit)

# Display the interface
print("Chatbot initialized! Type your message below and press Enter.")
display(output_area)
display(text_input)
```
