```python
import numpy as np
from scipy import stats
Height=[176,156,178,195,165,155,147,135,167,170,172,156,147,156,180,147,147]
height = np.array(Height)
Mean = np.mean(height)
Median = np.median(height)
Mode = stats.mode(height)
# print("Mean",Mean,"Median",Median,"Mode",Mode)
sd = np.std(height)
print("Standard deviation is :",sd)


var = np.var(height)
print(var)
std = np.sqrt(var)
print(std)
```

```python
import numpy as np
from scipy import stats

Age=[2,4,5,53,42,32,4,24,35,66,56,5,6,5,64,53,42,32,32,32,23]

speed = np.array([86,87,88,86,87,85,86])

age = np.array(Age)
Mean = np.mean(speed)
Median = np.median(speed)
Mode = stats.mode(speed)
sd = np.std(age)
var = np.var(age)
print("Mean, Median, Mode of age is : ",Mean,Median,Mode)
print("Standard Deviation Age is :",sd)
print("Variance of the Age is :",var)
```
